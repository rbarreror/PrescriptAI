```python
import json
import os

# 1. Definir la ruta al archivo JSON
# (Asumiendo que 'kaggle.json' está en la misma carpeta que tu notebook)
ruta_json = '../.kaggle/access_token.json'

# 2. Leer el archivo y cargar las credenciales como un diccionario
with open(ruta_json, 'r') as archivo:
    credenciales = json.load(archivo)

# 3. Inyectar las credenciales en las variables de entorno de tu sistema
os.environ['KAGGLE_USERNAME'] = credenciales['username']
os.environ['KAGGLE_API_TOKEN'] = credenciales['key']
```
